# Appendix D: Loading LFM3 weights

This edition uses a single candidate, `LiquidAI/LFM3-1.7B-Base`. Qwen configurations and checkpoints cannot be loaded into LFM3. The loader pins the official weights and tokenizer to a verified model revision.

```python
from reasoning_from_scratch.ch02 import get_device
from reasoning_from_scratch.lfm3 import (
    MODEL_ID, MODEL_REVISION, load_model_and_tokenizer,
)

print(MODEL_ID, MODEL_REVISION)
device = get_device()
model, tokenizer = load_model_and_tokenizer(device=device)
```

```python
import torch
from reasoning_from_scratch.ch02 import generate_text_basic_stream_cache

prompt = "The nematode C. elegans is"
input_ids = torch.tensor(tokenizer.encode(prompt), device=device).unsqueeze(0)
for token in generate_text_basic_stream_cache(
    model, input_ids, max_new_tokens=64, eos_token_id=tokenizer.eos_token_id,
):
    print(tokenizer.decode(token.squeeze(0).tolist()), end="", flush=True)
```
